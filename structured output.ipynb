```python
import anthropic

client = anthropic.Anthropic()

messages = []

# helper functions
def add_user_message(messages, content):
    messages.append({"role": "user", "content": content})


def chat(messages, system=None, temperature=0.0):
    params = {
        "model": "claude-sonnet-4-6",
        "max_tokens": 1024,
        "temperature": temperature,
        "messages": messages
    }

    if system is not None:
        params["system"] = system

    response = client.messages.create(**params)
    
    return response.content[0].text
```

```python
messages = []
add_user_message(messages, "Show me 3 AWS CLI command examples")
response = chat(messages, system="Output only raw bash commands, one per line. No markdown, no code fences, no explanation.")
print(response)
```
